## Carga del dataset

Se carga el archivo `diabetes.csv` desde la carpeta `data/raw/` del repositorio.
Esta carpeta contiene los datos originales sin modificar, siguiendo el principio
de preservación de la fuente primaria establecido en la arquitectura del proyecto.


```python
# ==============================================
# Carga del dataset mediante función documentada
# ==============================================

def cargar_datos(ruta):
    """
    Carga el dataset desde la ruta especificada.
    Parámetros:
        ruta (str): ruta relativa al archivo CSV.
    Retorna:
        pd.DataFrame: DataFrame con los datos cargados.
    """
    df = pd.read_csv(ruta)
    print(f"Dataset cargado correctamente ✓")
    print(f"Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas")
    return df

# Ejecución
df = cargar_datos('../data/raw/diabetes.csv')
```

```python
# ============================================================
# Inspección estructural mediante función modular documentada
# ============================================================

def inspeccionar_datos(df):
    """
    Ejecuta inspección estructural completa del dataset.
    Parámetros:
        df (pd.DataFrame): DataFrame a inspeccionar.
    Incluye:
        - Estructura general (.info)
        - Estadísticas descriptivas (.describe)
        - Valores nulos por columna (.isnull)
        - Balance de la variable objetivo (.value_counts)
    """
    print("=" * 50)
    print("1. ESTRUCTURA GENERAL")
    print("=" * 50)
    df.info()

    print("\n" + "=" * 50)
    print("2. ESTADÍSTICAS DESCRIPTIVAS")
    print("=" * 50)
    print(df.describe().round(2))

    print("\n" + "=" * 50)
    print("3. VALORES NULOS POR COLUMNA")
    print("=" * 50)
    print(df.isnull().sum())

    print("\n" + "=" * 50)
    print("4. BALANCE DE LA VARIABLE OBJETIVO")
    print("=" * 50)
    print(df['Diabetic'].value_counts())
    print(f"\nPorcentaje diabéticos: {df['Diabetic'].mean()*100:.1f}%")

# Ejecución
inspeccionar_datos(df)
```

---

## Interpretación de la inspección estructural

- **Sin valores nulos:** el dataset no presenta ausencias de datos, lo que
  simplifica el preprocesamiento inicial.
- **Tipos de datos coherentes:** todas las variables clínicas son numéricas,
  compatibles con los algoritmos de clasificación previstos para F3.
- **Variable objetivo:** `Diabetic` presenta distribución 67% / 33%,

- **PatientID:** es un identificador único sin valor predictivo,
